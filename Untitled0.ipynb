Q3.What is an object in OOps ?

Q4. what is diffrent between abstraction and encapulation ?

:


Ans: While both **Abstraction** and **Encapsulation** are fundamental principles of Object-Oriented Programming (OOP) and often go hand-in-hand, they serve different purposes:

**Encapsulation:**
*   **What it is:** Bundling the data (attributes) and the methods (functions) that operate on that data into a single unit (a class/object). It also restricts direct access to some of the object's components, meaning that internal state is hidden and protected from external misuse.
*   **Purpose:** To control access to an object's internal state and prevent unauthorized modification. It's about **data hiding** and **binding data with methods**.
*   **How:** Achieved using access modifiers (like `private`, `protected` in languages like Java/C++, or conventions like leading underscores in Python).
*   **Analogy:** Think of a car's engine. Encapsulation means all the engine parts (data) and how they work together (methods) are contained within the engine block. You can't just reach in and mess with individual pistons directly.

**Abstraction:**
*   **What it is:** Showing only the essential features of an object and hiding the complex implementation details. It focuses on *what* an object does rather than *how* it does it.
*   **Purpose:** To reduce complexity and improve efficiency by allowing users to interact with objects at a high level, without needing to know the intricate details of their inner workings. It's about **information hiding**.
*   **How:** Achieved through abstract classes, interfaces, and simply defining clear public methods that provide functionality without exposing internal logic.
*   **Analogy:** When you drive a car, you use the steering wheel, accelerator, and brake pedals (the abstract interface). You don't need to know the complex internal mechanisms of how the engine works, how the transmission shifts, or how the braking system operates to drive the car. Abstraction hides these complexities.

Q5.What is dunder method in python?

Ans: A dunder method (short for "double underscore" method) in Python is a special predefined method that starts and ends with two underscores, such as __init__ or __str__. They are also referred to as magic methods.

Dunder methods allow custom classes to emulate built-in behaviors, such as operator overloading, object initialization, sequence behavior, and custom string representations.

Common Dunder Methods

__init__(self, ...): Initializer method invoked automatically when a new instance of a class is created.

__str__(self): Defines a user-friendly string representation of an object (called by str() or print()).

__repr__(self): Defines an official, developer-focused string representation (called by repr()).

__len__(self): Returns the length of an object when len() is called on it.

__add__(self, other): Defines the behavior for the addition operator (+).

__eq__(self, other): Defines equality testing for the == operator.

Q7.What is Polymorphism in OOP ?

Polymorphism in Object-Oriented Programming (OOP) means "many forms." It is the ability of different classes to respond to the same method call in their own unique way, allowing a single interface to represent different underlying forms or data types.

It enables code flexibility and scalability by letting you treat objects of different types uniformly while executing their specific implementations.

Key Types of Polymorphism

Compile-Time Polymorphism (Static / Method Overloading): Multiple methods in the same class share the same name but have different parameters (number or types of arguments).

Run-Time Polymorphism (Dynamic / Method Overriding): A child class provides a specific implementation of a method that is already defined in its parent class. The method executed is determined dynamically at runtime based on the object instance.

1. Create a parent class Animal with a method speak() that prints a generic message. Create a child class Dog
that overrides the speak() method to print "Bark!"

```python
# Parent class
class Animal:
    def speak(self):
        print("Animals make sounds.")

# Child class
class Dog(Animal):
    def speak(self):
        print("Bark!")

# Create objects
animal = Animal()
dog = Dog()

# Call methods
animal.speak()  # Output: Animals make sounds.
dog.speak()     # Output: Bark!
```

2. Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectangle
from it and implement the area() method in both.

```python
from abc import ABC, abstractmethod
import math

# Abstract class
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

# Derived class Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

# Derived class Rectangle
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

# Create objects
c = Circle(5)
r = Rectangle(4, 6)

# Display areas
print("Area of Circle:", c.area())
print("Area of Rectangle:", r.area())
```

3. Implement a multi-level inheritance scenario where a class Vehicle has an attribute type. Derive a class Car
and further derive a class ElectricCar that adds a battery attribute

```python
# Base class
class Vehicle:
    def __init__(self, vehicle_type):
        self.type = vehicle_type

# Derived class from Vehicle
class Car(Vehicle):
    def __init__(self, vehicle_type, brand):
        super().__init__(vehicle_type)
        self.brand = brand

# Derived class from Car
class ElectricCar(Car):
    def __init__(self, vehicle_type, brand, battery):
        super().__init__(vehicle_type, brand)
        self.battery = battery

    def display(self):
        print("Type:", self.type)
        print("Brand:", self.brand)
        print("Battery:", self.battery)

# Create object
ev = ElectricCar("Electric Vehicle", "Tesla", "75 kWh")

# Display details
ev.display()
```

4. Demonstrate polymorphism by creating a base class Bird with a method fly(). Create two derived classes
Sparrow and Penguin that override the fly() method.

```python
# Base class
class Bird:
    def fly(self):
        print("Birds can fly.")

# Derived class Sparrow
class Sparrow(Bird):
    def fly(self):
        print("Sparrow flies high in the sky.")

# Derived class Penguin
class Penguin(Bird):
    def fly(self):
        print("Penguin cannot fly, but it can swim.")

# Polymorphism
birds = [Sparrow(), Penguin()]

for bird in birds:
    bird.fly()
```

5. Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes
balance and methods to deposit, withdraw, and check balance

```python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance  # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: ₹{amount}")
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrawn: ₹{amount}")
        else:
            print("Insufficient balance.")

    def check_balance(self):
        print(f"Current Balance: ₹{self.__balance}")


# Create account
account = BankAccount(1000)

# Perform operations
account.check_balance()
account.deposit(500)
account.withdraw(300)
account.check_balance()
```

6. Demonstrate runtime polymorphism using a method play() in a base class Instrument. Derive classes Guitar
and Piano that implement their own version of play()

```python
# Base class
class Instrument:
    def play(self):
        print("Instrument is being played.")

# Derived class Guitar
class Guitar(Instrument):
    def play(self):
        print("Guitar is strumming.")

# Derived class Piano
class Piano(Instrument):
    def play(self):
        print("Piano is playing melodies.")

# Runtime Polymorphism
instruments = [Guitar(), Piano()]

for instrument in instruments:
    instrument.play()
```

10. Demonstrate operator overloading by creating a class Vector and overriding the add method to add two
vectors

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading the + operator
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"


# Create vector objects
v1 = Vector(2, 3)
v2 = Vector(4, 5)

# Add vectors
v3 = v1 + v2

# Display result
print("Vector 1:", v1)
print("Vector 2:", v2)
print("Sum of Vectors:", v3)
```

14. Create a class Employee with a method calculate_salary() that computes the salary based on hours worked
and hourly rate. Create a derived class Manager that adds a bonus to the salary


```python
# Base class
class Employee:
    def __init__(self, hours_worked, hourly_rate):
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate


# Derived class
class Manager(Employee):
    def __init__(self, hours_worked, hourly_rate, bonus):
        super().__init__(hours_worked, hourly_rate)
        self.bonus = bonus

    def calculate_salary(self):
        return super().calculate_salary() + self.bonus


# Create objects
employee = Employee(40, 20)
manager = Manager(40, 20, 500)

# Display salaries
print("Employee Salary:", employee.calculate_salary())
print("Manager Salary:", manager.calculate_salary())
```

16. Create a class Animal with an abstract method sound(). Create two derived classes Cow and Sheep that
implement the sound() method

```python
from abc import ABC, abstractmethod

# Abstract class
class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

# Derived class Cow
class Cow(Animal):
    def sound(self):
        print("Moo!")

# Derived class Sheep
class Sheep(Animal):
    def sound(self):
        print("Baa!")

# Create objects
cow = Cow()
sheep = Sheep()

# Call methods
cow.sound()
sheep.sound()
```
