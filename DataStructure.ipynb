##Data Types and Structures Questions


Q1.What are data structures, and why are they important ?

Ans:Data structures are ways of organizing and storing data in a computer so it can be used efficiently. They are important because they help in faster searching, sorting, and managing of data, making programs work better and faster.

Q2.Explain the difference between mutable and immutable data types with examples?

Ans :Mutable data types are those whose values can be changed after creation, while immutable data types cannot be changed once created. For example, a list in Python is mutable because we can add, remove, or change its elements. On the other hand, a tuple and string are immutable because their values cannot be modified after they are created. Mutable types allow changes directly, while immutable types create a new object when modified.

Q3.What are the main differences between lists and tuples in Python?

Ans :Lists and tuples are both used to store multiple items in Python, but they have some important differences. A list is **mutable**, which means its elements can be changed, added, or removed after creation. A tuple is **immutable**, meaning its elements cannot be changed once created. Lists are written using square brackets `[ ]`, while tuples use round brackets `( )`. Lists are generally used when data may change, whereas tuples are used for fixed data that should remain constant.


Q4.Describe how dictionaries store data ?

A dictionary in Python stores data in **key-value pairs**. Each key is unique and is used to access its related value quickly. Dictionaries are written using curly braces `{ }`. For example, in `{"name": "Ankur", "age": 20}`, **“name”** and **“age”** are keys, while **“Ankur”** and **20** are values. Dictionaries help in organizing and retrieving data efficiently.



Q5.Why might you use a set instead of a list in Python?

Ans:A set is used instead of a list in Python when you want to store **unique values only** and avoid duplicates. Sets also allow faster searching and checking of elements compared to lists. For example, if you want to store unique student names without repetition, a set is a better choice than a list.


Q6.What is a string in Python, and how is it different from a list

A string in Python is a sequence of characters used to store text, such as names or sentences. It is written inside quotes (`" "` or `' '`). A string is **immutable**, which means its characters cannot be changed after creation. On the other hand, a list stores multiple items of different data types and is **mutable**, meaning its elements can be changed, added, or removed. Strings are mainly used for text, while lists are used for storing collections of data.


Q7.How do tuples ensure data integrity in Python

Ans:Tuples ensure data integrity in Python because they are **immutable**, meaning their values cannot be changed after creation. This prevents accidental modification of data and keeps information safe and consistent throughout the program. Tuples are useful when data should remain fixed, such as coordinates, dates, or constant values.


Q8.What is a hash table, and how does it relate to dictionaries in Python ?

Ans:A hash table is a data structure used to store data in **key-value pairs** for fast access and retrieval. In Python, dictionaries are based on **hash tables**, which help find values quickly using keys instead of searching through all data. This makes dictionaries efficient for storing and managing data.



Q9.Can lists contain different data types in Python ?

**ANS**.Yes, lists in Python can contain **different data types** in a single list. For example, a list can store integers, strings, floats, and even other lists together. Example: `[10, "Ankur", 5.5, True]`. This flexibility makes lists useful for storing mixed types of data.



Q10.Explain why strings are immutable in Python

Ans:Strings are immutable in Python, which means their values cannot be changed after creation. This helps improve security, memory efficiency, and performance because the same string can be reused without modification. If changes are needed, Python creates a new string instead of modifying the original one.


Q11.What advantages do dictionaries offer over lists for certain tasks

Ans:Dictionaries offer advantages over lists because they store data in **key-value pairs**, making it faster to search and access information using keys. In a list, data is accessed by index position, which may take more time for searching. Dictionaries are useful when data needs to be organized with names or labels, such as storing student details or employee records.


Q12.Describe a scenario where using a tuple would be preferable over a list

Ans:A tuple is preferable over a list when data should remain **fixed and unchanged**. For example, storing the coordinates of a location `(10, 20)` or days of the week, where values do not need modification. Since tuples are immutable, they help keep data safe and prevent accidental changes.


Q13.How do sets handle duplicate values in Python?

Ans:Sets in Python automatically **remove duplicate values** and store only unique elements. If the same value is added multiple times, it will appear only once in the set. For example, `{1, 2, 2, 3}` becomes `{1, 2, 3}`. This makes sets useful for storing unique data.



Q14.How does the “in” keyword work differently for lists and dictionaries?

Ans:The `in` keyword is used to check if an element exists, but it works differently for lists and dictionaries. In a **list**, `in` checks for values in the list. In a **dictionary**, `in` checks for **keys**, not values. For example, `"apple" in ["apple", "banana"]` returns **True**, while `"name" in {"name": "Ankur"}` checks if the key exists.



Q15.Can you modify the elements of a tuple? Explain why or why not ?

Ans.No, you cannot modify the elements of a tuple because tuples are **immutable** in Python. Once a tuple is created, its values cannot be changed, added, or removed. This helps protect data from accidental changes and ensures data integrity.


Q16.What is a nested dictionary, and give an example of its use case?

Ans.A nested dictionary is a dictionary that contains **another dictionary inside it**. It is used to store complex or grouped data in an organized way. For example, a school system can use a nested dictionary to store student details like name, age, and marks: `{"student1": {"name": "Ankur", "age": 20}}`. This helps in managing related data efficiently.



Q17.Describe the time complexity of accessing elements in a dictionary

Ans.The time complexity of accessing elements in a dictionary is usually **O(1)**, which means it takes **constant time**. This is because dictionaries use a **hash table** to quickly find values using keys. However, in rare cases, it can become **O(n)** if many keys have the same hash value. Dictionaries are generally very fast for searching and retrieving data.


Q18.In what situations are lists preferred over dictionaries?

Lists are preferred over dictionaries when data needs to be stored in an **ordered sequence** and accessed by **index position**. They are useful when items do not need keys, such as storing marks, names, or a collection of similar data. Lists are better for simple data storage and when the order of elements is important.


Q19.Why are dictionaries considered unordered, and how does that affect data retrieval

Ans.Dictionaries are considered unordered because they store data as **key-value pairs** instead of fixed positions like lists. Data is not accessed by index number but by using **keys**. This makes data retrieval faster and more efficient, as Python can quickly find values through keys without checking every element.


Q20.Explain the difference between a list and a dictionary in terms of data retrieval?

Ans.A list and a dictionary differ in how data is retrieved. In a **list**, data is accessed using an **index number** (position), such as `list[0]`. In a **dictionary**, data is accessed using a **key**, such as `dict["name"]`. Dictionaries are generally faster for searching specific data, while lists are useful for ordered collections of items.
