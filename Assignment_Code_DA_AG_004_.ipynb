# Question 1 : What is a RESTful API?

A **RESTful API** (Representational State Transfer) is an architectural style for designing networked applications. It relies on a stateless, client-server communication protocol—almost always **HTTP**.

### Key Principles of REST:

1. **Client-Server Architecture**:
   The client (user interface) and the server (data storage and processing) are managed independently. This separation allows both components to evolve without relying on one another.

2. **Statelessness**:
   Each request from a client to the server must contain all the information necessary to understand and process the request. The server does not store any session context about the client.

3. **Cacheability**:
   Responses must define themselves as cacheable or non-cacheable to improve network efficiency and scalability.

4. **Uniform Interface**:
   To simplify the architecture, REST uses a standardized way of interacting with servers, typically involving:
   * **Resources**: Identified in requests using URIs (Uniform Resource Identifiers) like `/api/v1/users`.
   * **HTTP Methods**: Standard verbs to perform actions:
     * `GET`: Retrieve a resource.
     * `POST`: Create a new resource.
     * `PUT` / `PATCH`: Update an existing resource.
     * `DELETE`: Remove a resource.
   * **Representation**: Resources can be represented in different formats, most commonly **JSON** or **XML**.

# Question 2: What is Flask, and why is it popular for building APIs?

**Flask** is a micro web framework written in Python. Unlike "batteries-included" frameworks like Django, Flask does not require a particular directory structure, database integration, or form validation tools by default. It provides the bare essentials needed to build web applications and APIs, leaving structural and library choices to the developer.

### Why Flask is popular for building APIs:

1. **Minimalist and Lightweight (Microframework)**:
   Flask has a small footprint and minimal dependencies, making it fast to set up and highly performant for simple API endpoints.

2. **Flexibility and Control**:
   Because it doesn't force design decisions on the developer, you have complete control over which databases, ORMs, authentication methods, and libraries to integrate.

3. **Easy to Learn and Use**:
   Its routing system is highly intuitive. You can set up a basic, working API in just a few lines of Python code using decorators like `@app.route()`.

4. **Extensive Ecosystem**:
   Despite being a microframework, it has a rich ecosystem of extensions (e.g., `Flask-RESTful`, `Flask-SQLAlchemy`, `Flask-JWT-Extended`) that make scaling and adding advanced API features simple.

5. **Great for Microservices and Prototyping**:
   Its lightweight nature makes it an excellent choice for microservices architectures, serverless deployments, and rapid prototyping.

# Question 3: What are HTTP methods used in RESTful APIs? (Detailed Explanation)

In RESTful web services, HTTP methods (often called **verbs**) map directly to database **CRUD (Create, Read, Update, Delete)** operations. To make an API truly RESTful, these standard methods must be used in accordance with their defined specifications in the HTTP protocol.

Here is an in-depth breakdown of the primary HTTP methods, their design contracts, safety, idempotency, and common response codes:

---

### 1. GET (Read / Retrieve)
* **Purpose**: Used to request and retrieve data from a specified resource. It must only fetch data and have no side effects on the server state.
* **Properties**:
  * **Safe**: Yes (It does not modify resource state).
  * **Idempotent**: Yes (Making multiple identical requests yields the same result without changing anything on the server).
* **Usage Scenarios**:
  * `GET /api/v1/users` – Retrieve a list of all users.
  * `GET /api/v1/users/42` – Retrieve detailed profile data for the user with ID `42`.
* **Common Response Codes**:
  * `200 OK`: Request succeeded with the requested data returned in the payload.
  * `404 Not Found`: The requested resource ID does not exist.

---

### 2. POST (Create)
* **Purpose**: Used to submit data to the server to create a new resource. The server processes the request payload (typically JSON) and decides where the new resource lives.
* **Properties**:
  * **Safe**: No (It changes the server state by creating a new entry).
  * **Idempotent**: No (Sending the same POST request multiple times will typically create multiple identical resources, resulting in duplicates).
* **Usage Scenario**:
  * `POST /api/v1/users` with payload `{"name": "Alice", "email": "alice@example.com"}` – Creates a new user profile on the server.
* **Common Response Codes**:
  * `201 Created`: The resource was successfully created. A `Location` header pointing to the URI of the new resource is often returned.
  * `400 Bad Request`: The sent payload was malformed, or missing required attributes.

---

### 3. PUT (Replace / Full Update)
* **Purpose**: Used to update an existing resource or create it if it doesn't already exist at a *specific* URI. A PUT request requires sending the **entire** updated resource payload. If you omit any fields, the server is expected to overwrite those missing fields with null or default values.
* **Properties**:
  * **Safe**: No (It alters server-side data).
  * **Idempotent**: Yes (Replacing a resource with the exact same payload multiple times will result in the same final state on the server).
* **Usage Scenario**:
  * `PUT /api/v1/users/42` with payload `{"name": "Alice Smith", "email": "alice.smith@example.com"}` – Overwrites the entire user object with ID `42`.
* **Common Response Codes**:
  * `200 OK` or `204 No Content`: The resource was successfully updated.
  * `201 Created`: If the resource did not exist and PUT successfully created a new one.

---

### 4. PATCH (Partial Update)
* **Purpose**: Used to apply partial modifications to a resource. Unlike `PUT`, you only send the specific fields you wish to change instead of the whole payload. This saves network bandwidth and reduces database overwrite conflicts.
* **Properties**:
  * **Safe**: No.
  * **Idempotent**: No (Though in practice, many simple PATCH implementations act idempotently, the specification does not guarantee it. For instance, a PATCH request that increments an integer field is non-idempotent).
* **Usage Scenario**:
  * `PATCH /api/v1/users/42` with payload `{"email": "alice.new@example.com"}` – Updates only the email field of user `42`, leaving the name and other parameters untouched.
* **Common Response Codes**:
  * `200 OK` or `204 No Content`: The partial update was completed successfully.

---

### 5. DELETE (Remove)
* **Purpose**: Used to remove a specific resource from the server.
* **Properties**:
  * **Safe**: No (It modifies the state by deleting data).
  * **Idempotent**: Yes (The first successful DELETE removes the resource. Subsequent DELETE requests will still result in the resource being deleted, though the server may return a different status code like `404 Not Found`).
* **Usage Scenario**:
  * `DELETE /api/v1/users/42` – Permanently deletes the user with ID `42`.
* **Common Response Codes**:
  * `200 OK` or `204 No Content`: The deletion was successfully executed.
  * `404 Not Found`: The resource has already been deleted or never existed.

---

### Summary Table of RESTful HTTP Methods

| HTTP Method | CRUD Operation | Safe? | Idempotent? | Success Status Codes |
| :--- | :--- | :--- | :--- | :--- |
| **GET** | Read | **Yes** | **Yes** | 200 OK |
| **POST** | Create | No | No | 201 Created |
| **PUT** | Update (Replace) | No | **Yes** | 200 OK / 204 No Content |
| **PATCH** | Update (Modify) | No | No | 200 OK / 204 No Content |
| **DELETE** | Delete | No | **Yes** | 200 OK / 204 No Content / 404 Not Found |

# Question 4: What is the purpose of the `@app.route()` decorator in Flask?

In Flask, the `@app.route()` decorator is one of the most fundamental tools. Its primary purpose is to **bind (or map) a specific URL path to a Python function**. This process is known as **routing**.

When a user or a client application sends an HTTP request to your Flask server, Flask uses these route definitions to determine which function (known as a **view function**) should execute to process the request and return a response.

---

### 1. Basic Routing Mechanism

At its simplest, the decorator takes a string representing the URL path as its first argument. Here is a basic example:

```python
@app.route('/hello')
def say_hello():
    return "Hello, World!"
```

* **The Decorator (`@app.route('/hello')`)**: Instructs Flask that any incoming browser or API request for the URL path `/hello` should be directed to the function immediately below it.
* **The View Function (`say_hello()`)**: Processes the request and returns a response (in this case, a simple text string, but it could be HTML, JSON, or a redirect).

---

### 2. Handling Dynamic Routes (URL Variables)

APIs rarely deal only with static URLs. You often need to capture variable segments from the URL path—such as a specific database ID. Flask allows you to define dynamic routes using variable rules marked by `<variable_name>` syntax. You can also specify an optional converter to restrict the type of data accepted:

```python
@app.route('/api/v1/users/<int:user_id>')
def get_user_profile(user_id):
    # The 'user_id' from the URL is automatically passed as an argument
    return f"Fetching profile for user #{user_id}"
```

**Common Converters**:
* `<string>`: Accepts any text without a slash (default).
* `<int>`: Accepts positive integers.
* `<float>`: Accepts positive floating-point values.
* `<path>`: Like a string, but also accepts slashes.
* `<uuid>`: Accepts RFC 4122 UUID strings.

---

### 3. Restricting Route Access by HTTP Methods

By default, a route defined with `@app.route()` only listens to **GET** requests. If you want a route to handle other HTTP methods (like `POST`, `PUT`, or `DELETE`), you must explicitly declare them using the `methods` argument:

```python
@app.route('/api/v1/items', methods=['GET', 'POST'])
def handle_items():
    from flask import request
    if request.method == 'POST':
        # Code to handle item creation
        return "Item created", 201
    
    # Code to handle listing items
    return "Here are your items", 200
```

This behavior makes it easy to structure a clean, resource-oriented endpoint interface.

---

### 4. How It Works Under the Hood

In Python, a **decorator** is syntactical sugar. Writing:

```python
@app.route('/')
def index():
    return "Home"
```

is equivalent to calling Flask's explicit route-registering method:

```python
def index():
    return "Home"

app.add_url_rule('/', endpoint='index', view_func=index)
```

The `@app.route()` decorator simply automates this behind-the-scenes registration process, keeping your codebase clean, highly readable, and declarative.
